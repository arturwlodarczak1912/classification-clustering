```python
import pickle
```

```python
with open('drug_tree_model.pkl', 'rb') as f:
    clf, ohe = pickle.load(f)
```

```python
import pandas as pd
```

```python
df = pd.read_excel('dane do predykcji.xlsx')
```

```python
df
```

## Kod predykcyjny

```python
import numpy as np
```

```python
X = df
```

```python
X_cat = X.select_dtypes(include='object')
X_num = X.select_dtypes(include='number')
```

```python
X_one = ohe.transform(X_cat)
```

```python
X = np.concatenate((X_num.to_numpy(), X_one), axis=1)
```

```python
clf.predict(X)
```

```python
df['Drug_prediction'] = clf.predict(X)
```

```python
df.to_excel('dane_i_predykcje.xlsx', index=False)
```
